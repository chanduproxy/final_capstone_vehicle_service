```python
vehicle_bronze = spark.read.format("delta").table("final_capstone_project.bronze.vehicle_bronze")
vehicle_bronze.display()
```

```python
from pyspark.sql.functions import upper, col

vehicle_silver = (
    vehicle_bronze
    .withColumn("vehicle_type", upper(col("vehicle_type")))
    .withColumn("service_type", upper(col("service_type")))
    .withColumn("service_city", upper(col("service_city")))
    .withColumn("service_amount", col("labour_charge") + col("parts_charge") - col("discount"))
)
```

```python
display(vehicle_silver)
```

```python
vehicle_silver.write\
    .format("delta")\
    .mode("overwrite")\
    .option("overwriteSchema", "true")\
    .option("header","true")\
    .saveAsTable("final_capstone_project.silver.vehicle_silver")
```
