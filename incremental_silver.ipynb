```python
vehicle_bronze_increment = spark.read.format("delta").table("final_capstone_project.bronze.vehicle_bronze_increment")
vehicle_bronze_increment.display()
```

```python
from pyspark.sql.functions import upper, col

vehicle_silver_increment = (
    vehicle_bronze_increment
    .withColumn("vehicle_type", upper(col("vehicle_type")))
    .withColumn("service_type", upper(col("service_type")))
    .withColumn("service_city", upper(col("service_city")))
    .withColumn("service_amount", col("labour_charge") + col("parts_charge") - col("discount"))
)
```

```python
display(vehicle_silver_increment)
```

```python
vehicle_silver_increment.write\
    .format("delta")\
    .mode("overwrite")\
    .option("overwriteSchema", "true")\
    .option("header","true")\
    .saveAsTable("final_capstone_project.silver.vehicle_silver_incremental")
```
